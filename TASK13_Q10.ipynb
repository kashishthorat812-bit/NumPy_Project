```python
import numpy as np

try:
    n = int(input("How many numbers do you want to generate? "))

    if n <= 0:
        print("Please enter a positive number.")
    else:
        arr = np.random.randint(10, 101, n)

        print("\nGenerated Array:")
        print(arr)

        print("\n<<----Statastics---->>")
        print("Mean:", np.mean(arr))
        print("Median:", np.median(arr))
        print("Standard Deviation:", np.std(arr))
        print("Minimum:", np.min(arr))
        print("Maximum:", np.max(arr))

        if n % 2 == 0:
            arr2d = arr.reshape(2, n // 2)
            print("\nReshaped 2D Array:")
            print(arr2d)

            print("\nRow-wise Sum:")
            print(np.sum(arr2d, axis=1))
        else:
            print("\nCannot reshape into 2D array equally because number count is odd.")

except ValueError:
    print("Invalid input! Please enter an integer.")
```
