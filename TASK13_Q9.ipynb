```python
import numpy as np
# create a 1D array of 20 random integers
a=np.random.randint(1,50,20)
print("1D Array :")
print(a)

# reshape the 1D array into 4*5 matrix
matrix=a.reshape(4,5)
print("\n4*5 Matrix :")
print(matrix)

# sum ,mean and standard Deviation of matrix
print("\n<<----Statistics---->>")
print("Sum of matrix = ",np.sum(matrix))
print("Mean of matrix = ",np.mean(matrix))
print("Standard deviation of matrix = ",np.std(matrix))

# maximum value in each row
print("\nMaximum value in each row :")
print(np.max(matrix,axis=1))
```
