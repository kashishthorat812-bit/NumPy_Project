```python
import numpy as np
# print Random 10 numbers between 0 and 1 using rand
print("Random 10 numbers between 0 and 1 :")
a=np.random.rand(10)
print(a)

# print Random number of matrix 3*3 using randn
a2=np.random.randn(3,3)
print("\nRandom number of matrix 3*3")
print(a2)

# print 2D Array of random integers between 10 to 50 with matrix 4*5 using randint
a3=np.random.randint(10,51,(4,5))
print("\n2D Array of random integers between 10 to 50 with matrix 4*5 :")
print(a3)
```
