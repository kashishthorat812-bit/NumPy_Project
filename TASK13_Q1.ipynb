```python
import numpy as np
# create a 1D Array
print("1D Array :")
a=np.array([10,20,30,40,50])
print(a)
print("Shape = ",a.shape)
print("Data type = ",a.dtype)

# create a 2D Array
print("\n2D array :")
a1=np.array([[10,20,30],[40,50,60],[70,80,90]])
print(a1)
print("Shape = ",a1.shape)
print("Data type = ",a1.dtype)
```
