```python
import numpy as np

# print 10 equally spaced numbers between 0 and 5
a = np.linspace(0, 5, 10)
print("10 numbers between 0 and 5:")
print(a)
print("Length:", len(a))

# print 15 equally spaced numbers between -10 and 10
a1 = np.linspace(-10, 10, 15)
print("\n15 numbers between -10 and 10:")
print(a1)
print("Length:", len(a1))
```
