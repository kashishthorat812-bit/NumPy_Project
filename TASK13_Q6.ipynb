```python
import numpy as np
v1=np.array([2,4,6,8])
v2=np.array([1,3,5,7])

# Addition of two vector
add=v1+v2
print("Addition :")
print(add)

# Substraction of two vector
sub=v1-v2
print("\nSubstraction :")
print(sub)

# Multiplication of two vector
mul=v1*v2
print("\nElement-wise Multiplication :")
print(mul)

mul2=np.dot(v1,v2)
print("\nDot Product :")
print(mul2)
```
